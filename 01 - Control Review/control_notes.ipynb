```python
def squareLoop(input_int):
    theSquareSum = 0
    for current_int in range(1,input_int):
        theSquareSum = theSquareSum + current_int ** 2
    return theSquareSum
```

```python
squareLoop(5)
```

```python
def square_even(input_int):
    the_sum = 0
    current_int = 2
    while current_int < input_int:
        the_sum += current_int**2
        current_int += 2
    return the_sum

square_even(6)
```
