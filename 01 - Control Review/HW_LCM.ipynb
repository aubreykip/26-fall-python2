```python
def lcm(int1,int2):
    if int1 > int2:
        bigger_int = int1
    else:
        bigger_int = int2
    multiple = bigger_int
    while multiple % int1 != 0 or multiple % int2 != 0:
        multiple += bigger_int
    return multiple
```

```python
lcm(6,10)
```

```python
lcm(3,10)
```

```python
lcm(10,30)
```
